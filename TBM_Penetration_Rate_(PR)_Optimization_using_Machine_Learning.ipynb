# 1 — IMPORT LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
```

# 2 — load or READ DATASET

```python
df = pd.read_csv("/content/machine_failure_dataset.csv")
```

## Data Preprocessing

Steps:
1. Missing values handle karna
2. Noise reduce karna
3. Feature scaling
4. Relevant features select karna


```python
# Missing values ko mean se fill karna
df.fillna(df.mean(), inplace=True)
```

```python
# Feature selection
features = [
    'Advance_Time',
    'Excavation_per_Stroke',
    'Boring_Energy',
    'RPM',
    'Torque',
    'Thrust_Force',
    'Speed',
    'Gripper_Pressure',
    'Q_Value'
]
```

```python
X = df[features]

# Weak supervision ke liye PR ko categorical bana rahe hain
# (Low, Medium, High)
df['PR_Class'] = pd.qcut(df['Penetration_Rate'], q=3, labels=[0,1,2])
y = df['PR_Class']

# Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Weak Supervision Strategy

Real industrial data me exact labels hamesha available nahi hotay.
Is project me:

- Penetration Rate ko continuous value se
- Quantile-based categories me convert kiya gaya

This approach allows learning from noisy and approximate labels.


# 5 — TRAIN TEST SPLIT

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

## Machine Learning Model

Random Forest Classifier use kiya gaya kyun ke:
- Non-linear data handle karta hai
- Noise ke against robust hota hai
- Industrial datasets ke liye suitable hai


# 6 — TRAIN MODEL

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

## Model Evaluation

Hum following metrics use karte hain:
- Accuracy
- Precision
- Recall
- Confusion Matrix


# 7 — METRICS & CONFUSION MATRIX

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

## Results Interpretation

- Model ne Penetration Rate categories ko achi accuracy ke sath predict kiya
- Random Forest ne noisy TBM data me stable performance di
- Confusion matrix se pata chalta hai ke medium aur high PR classes me thori confusion hai
