# Smart Dosage Calculator –

```python
import tkinter as tk
from tkinter import messagebox

# Drug database - Added a placeholder for 'brand_name'
drug_database = {
    "paracetamol": {"dose_per_kg": 15, "max_per_kg": 60, "brand_name": "Calpol, Panadol"},
    "ibuprofen": {"dose_per_kg": 10, "max_per_kg": 40, "brand_name": "Brufen, Nurofen"},
    "amoxicillin": {"dose_per_kg": 20, "max_per_kg": 90, "brand_name": "Amoxil, Moxatag"},
    "cefixime": {"dose_per_kg": 8, "max_per_kg": 16, "brand_name": "Suprax, Taxim O"}
}

def calculate_dose_console():
    while True:
        drug_name = input("Enter Drug Name (or 'exit' to quit): ").lower()
        if drug_name == 'exit':
            break

        try:
            weight = float(input("Enter Weight (kg): "))
            tablet_strength = float(input("Enter Tablet Strength (mg): "))
        except ValueError:
            print("Error: Please enter valid numeric values for weight and tablet strength.")
            continue

        if drug_name in drug_database:
            dose_info = drug_database[drug_name]
            dose_per_kg = dose_info["dose_per_kg"]
            max_per_kg = dose_info["max_per_kg"]
            brand_name = dose_info.get("brand_name", "Not specified")

            single_dose = dose_per_kg * weight
            max_daily = max_per_kg * weight

            if tablet_strength <= 0:
                print("Error: Tablet strength must be greater than zero.")
                continue

            tablets_needed = single_dose / tablet_strength

            print(
                f"\nDrug: {drug_name.capitalize()}\n"
                f"Brand Names (Pakistan): {brand_name}\n"
                f"Single Dose: {round(single_dose,2)} mg\n"
                f"Max Daily Dose: {round(max_daily,2)} mg\n"
                f"Tablets Per Dose: {round(tablets_needed,2)}"
            )
        else:
            print(f"Warning: Drug '{drug_name}' not found in database. Please add it or try another drug.\n")

# Call the interactive function
calculate_dose_console()
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Drug database with brand names for Pakistan
drug_database = {
    "paracetamol": {"dose_per_kg": 15, "max_per_kg": 60, "brand_name": "Calpol, Panadol"},
    "ibuprofen": {"dose_per_kg": 10, "max_per_kg": 40, "brand_name": "Brufen, Nurofen"},
    "amoxicillin": {"dose_per_kg": 20, "max_per_kg": 90, "brand_name": "Amoxil, Moxatag"},
    "cefixime": {"dose_per_kg": 8, "max_per_kg": 16, "brand_name": "Suprax, Taxim O"}
}

def calculate_dose_interactive(drug_name_input, weight_input, tablet_strength_input):
    clear_output(wait=True)
    print("Smart Dosage Calculator\n")

    drug_name = drug_name_input.lower()

    try:
        weight = float(weight_input)
        tablet_strength = float(tablet_strength_input)
    except ValueError:
        print("Error: Please enter valid numeric values for weight and tablet strength.")
        return

    if drug_name in drug_database:
        dose_info = drug_database[drug_name]
        dose_per_kg = dose_info["dose_per_kg"]
        max_per_kg = dose_info["max_per_kg"]
        brand_name = dose_info.get("brand_name", "Not specified")

        single_dose = dose_per_kg * weight
        max_daily = max_per_kg * weight

        if tablet_strength <= 0:
            print("Error: Tablet strength must be greater than zero.")
            return

        tablets_needed = single_dose / tablet_strength

        print(
            f"Drug: {drug_name.capitalize()}\n"
            f"Brand Names (Pakistan): {brand_name}\n"
            f"Single Dose: {round(single_dose,2)} mg\n"
            f"Max Daily Dose: {round(max_daily,2)} mg\n"
            f"Tablets Per Dose: {round(tablets_needed,2)}"
        )
    else:
        print(f"Warning: Drug '{drug_name}' not found in database. Please add it or try another drug.")


# Create interactive widgets
drug_name_widget = widgets.Text(
    value='',
    placeholder='e.g., paracetamol',
    description='Drug Name:',
    disabled=False
)

weight_widget = widgets.FloatText(
    value=0.0,
    description='Weight (kg):',
    disabled=False
)

tablet_strength_widget = widgets.FloatText(
    value=0.0,
    description='Tablet Strength (mg):',
    disabled=False
)

button = widgets.Button(description="Calculate Dose")
output_area = widgets.Output()

def on_button_click(b):
    with output_area:
        calculate_dose_interactive(drug_name_widget.value, weight_widget.value, tablet_strength_widget.value)

button.on_click(on_button_click)

# Display the widgets
display(drug_name_widget, weight_widget, tablet_strength_widget, button, output_area)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Sample Medical Knowledge Dataset (from previous cells)
data = {
    "symptoms": [
        "fever cough cold",
        "headache nausea vomiting",
        "stomach pain diarrhea",
        "sore throat fever",
        "body pain fatigue",
        "chest pain breathing difficulty"
    ],
    "disease": [
        "Viral Flu",
        "Migraine",
        "Food Infection",
        "Throat Infection",
        "Dengue Risk",
        "Heart / Lung Emergency"
    ],
    "treatment": [
        "Rest + Paracetamol + Fluids",
        "Rest + Pain relief + Dark room",
        "ORS + Hydration + Light food",
        "Warm water gargle + Doctor visit",
        "Blood test + Hospital checkup",
        "Emergency hospital visit immediately"
    ]
}

df_medical = pd.DataFrame(data) # Renamed to avoid conflict with drug calculator df

# ML Model Training
vectorizer_medical = CountVectorizer() # Renamed to avoid conflict
X_medical = vectorizer_medical.fit_transform(df_medical["symptoms"])

model_medical = MultinomialNB() # Renamed to avoid conflict
model_medical.fit(X_medical, df_medical["disease"])

# AI Chat Response Generator function (adapted for interactive use)
def generate_response_interactive(user_input_symptoms):
    user_input_symptoms = user_input_symptoms.lower()

    # Disease Prediction
    X_test_medical = vectorizer_medical.transform([user_input_symptoms])
    prediction = model_medical.predict(X_test_medical)[0]

    # Treatment matching
    idx = df_medical["disease"].tolist().index(prediction)
    treatment = df_medical["treatment"][idx]

    responses = [
        f"🧠 AI Analysis → {prediction}",
        f"💊 Suggested Care → {treatment}",
        "⚠ If symptoms worsen, consult doctor immediately"
    ]
    return "\n".join(responses)

# Create interactive widgets
symptoms_widget = widgets.Textarea(
    value='',
    placeholder='Describe your symptoms here (e.g., fever, cough, cold)',
    description='Symptoms:',
    disabled=False,
    layout=widgets.Layout(width='500px', height='100px')
)

predict_button = widgets.Button(description="Predict Disease")
medical_output_area = widgets.Output()

def on_predict_button_click(b):
    with medical_output_area:
        clear_output(wait=True)
        if symptoms_widget.value:
            response = generate_response_interactive(symptoms_widget.value)
            print(response)
        else:
            print("Please enter your symptoms.")

predict_button.on_click(on_predict_button_click)

print("🤖 LEGENDARY AI MEDICAL ASSISTANT (Interactive)\n")
# Display the widgets
display(symptoms_widget, predict_button, medical_output_area)
```
