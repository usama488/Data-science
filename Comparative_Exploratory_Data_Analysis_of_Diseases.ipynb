# 5. Statistical Testing

# 5.1 T‑test for Feature Differences Between Classes

```python
# Breast Cancer: select a few features
bc_features_to_test = ['mean radius', 'mean texture', 'mean perimeter', 'mean area']
ttest_bc_results = []
for col in bc_features_to_test:
    group1 = df_bc[df_bc['target']==0][col]
    group2 = df_bc[df_bc['target']==1][col]
    t_stat, p_val = ttest_ind(group1, group2, equal_var=False)
    ttest_bc_results.append({'Feature': col, 'T-statistic': t_stat, 'P-value': p_val})
pd.DataFrame(ttest_bc_results)
```

```python
# Parkinson's Disease: select a few features
pd_features_to_test = ['MDVP:Fo(Hz)', 'MDVP:Fhi(Hz)', 'MDVP:Flo(Hz)', 'MDVP:Jitter(%)']
ttest_pd_results = []
for col in pd_features_to_test:
    group1 = df_pd[df_pd['target']==0][col]
    group2 = df_pd[df_pd['target']==1][col]
    t_stat, p_val = ttest_ind(group1, group2, equal_var=False)
    ttest_pd_results.append({'Feature': col, 'T-statistic': t_stat, 'P-value': p_val})
pd.DataFrame(ttest_pd_results)
```

# 6. Machine Learning (Classification)

# 6.1 Prepare Data

```python
# Breast Cancer
X_bc = df_bc.drop(['target', 'target_label'], axis=1)
y_bc = df_bc['target']
X_bc_train, X_bc_test, y_bc_train, y_bc_test = train_test_split(X_bc, y_bc, test_size=0.2, random_state=42, stratify=y_bc)
scaler_bc_ml = StandardScaler()
X_bc_train_scaled = scaler_bc_ml.fit_transform(X_bc_train)
X_bc_test_scaled = scaler_bc_ml.transform(X_bc_test)
```

```python
# Parkinson's Disease
X_pd = df_pd.drop('target', axis=1)
y_pd = df_pd['target']
X_pd_train, X_pd_test, y_pd_train, y_pd_test = train_test_split(X_pd, y_pd, test_size=0.2, random_state=42, stratify=y_pd)
scaler_pd_ml = StandardScaler()
X_pd_train_scaled = scaler_pd_ml.fit_transform(X_pd_train)
X_pd_test_scaled = scaler_pd_ml.transform(X_pd_test)
```

# 6.2 Define Models

```python
models = {
    'Logistic Regression': LogisticRegression(),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(n_estimators=100),
    'Gradient Boosting': GradientBoostingClassifier(),
    'SVM': SVC(probability=True)
}
```

# 6.3 Train & Evaluate on Breast Cancer

```python
results_bc = {}
for name, model in models.items():
    model.fit(X_bc_train_scaled, y_bc_train)
    y_pred = model.predict(X_bc_test_scaled)
    acc = accuracy_score(y_bc_test, y_pred)
    auc = roc_auc_score(y_bc_test, model.predict_proba(X_bc_test_scaled)[:,1])
    results_bc[name] = {'Accuracy': acc, 'AUC': auc}
    print(f"Breast Cancer - {name:20} | Accuracy: {acc:.4f} | AUC: {auc:.4f}")
```

# 6.4 Train & Evaluate on Parkinson's Disease

```python
results_pd = {}
for name, model in models.items():
    model.fit(X_pd_train_scaled, y_pd_train)
    y_pred = model.predict(X_pd_test_scaled)
    acc = accuracy_score(y_pd_test, y_pred)
    auc = roc_auc_score(y_pd_test, model.predict_proba(X_pd_test_scaled)[:,1])
    results_pd[name] = {'Accuracy': acc, 'AUC': auc}
    print(f"Parkinson's Disease - {name:20} | Accuracy: {acc:.4f} | AUC: {auc:.4f}")
```

# 6.5 Model Comparison Bar Charts

```python
res_bc_df = pd.DataFrame(results_bc).T.reset_index().rename(columns={'index': 'Model'})
res_bc_df['Dataset'] = 'Breast Cancer'
res_pd_df = pd.DataFrame(results_pd).T.reset_index().rename(columns={'index': 'Model'})
res_pd_df['Dataset'] = 'Parkinson\'s Disease'
combined = pd.concat([res_bc_df, res_pd_df])

fig = px.bar(combined, x='Model', y='Accuracy', color='Dataset', barmode='group',
             title='Model Accuracy Comparison')
fig.show()
```

# 7. Deep Learning (Neural Networks)

# 7.1 Breast Cancer Neural Network

```python
nn_bc = Sequential([
    Dense(128, activation='relu', input_shape=(X_bc_train_scaled.shape[1],)),
    BatchNormalization(),
    Dropout(0.3),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])
nn_bc.compile(optimizer=Adam(0.001), loss='binary_crossentropy', metrics=['accuracy'])
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history_bc = nn_bc.fit(X_bc_train_scaled, y_bc_train, epochs=100, batch_size=16, validation_split=0.2, callbacks=[early_stop], verbose=0)
loss_bc, acc_bc = nn_bc.evaluate(X_bc_test_scaled, y_bc_test, verbose=0)
print(f"Breast Cancer Neural Network Test Accuracy: {acc_bc:.4f}")
```

# 7.2 Parkinson's Disease Neural Network

```python
nn_pd = Sequential([
    Dense(64, activation='relu', input_shape=(X_pd_train_scaled.shape[1],)),
    BatchNormalization(),
    Dropout(0.3),
    Dense(32, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])
nn_pd.compile(optimizer=Adam(0.001), loss='binary_crossentropy', metrics=['accuracy'])
history_pd = nn_pd.fit(X_pd_train_scaled, y_pd_train, epochs=100, batch_size=16, validation_split=0.2, callbacks=[early_stop], verbose=0)
loss_pd, acc_pd = nn_pd.evaluate(X_pd_test_scaled, y_pd_test, verbose=0)
print(f"Parkinson's Disease Neural Network Test Accuracy: {acc_pd:.4f}")
```

```python
# ======================================================
# RUNTIME PREDICTION CELL: Compare Breast Cancer & Parkinson's
# ======================================================

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# -------------------------------
# 1. Load or train models (if not already saved)
# -------------------------------
def get_models():
    """Load or train breast cancer and Parkinson's models."""
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.preprocessing import StandardScaler
    from sklearn.datasets import load_breast_cancer
    import pandas as pd

    # ----- Breast Cancer model -----
    try:
        # Try to load saved model
        import joblib
        bc_model = joblib.load('bc_model.pkl')
        bc_scaler = joblib.load('bc_scaler.pkl')
        bc_features = joblib.load('bc_features.pkl')
        print(" Loaded saved Breast Cancer model.")
    except:
        print(" Training Breast Cancer model on‑the‑fly...")
        data = load_breast_cancer()
        X = pd.DataFrame(data.data, columns=data.feature_names)
        y = data.target
        from sklearn.model_selection import train_test_split
        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X)
        model = RandomForestClassifier(n_estimators=100, random_state=42)
        model.fit(X_scaled, y)
        bc_model = model
        bc_scaler = scaler
        bc_features = list(X.columns)
        # Optionally save for next time
        # joblib.dump(model, 'bc_model.pkl')
        # joblib.dump(scaler, 'bc_scaler.pkl')
        # joblib.dump(bc_features, 'bc_features.pkl')
        print(" Breast Cancer model trained.")

    # ----- Parkinson's Disease model -----
    try:
        import joblib
        pd_model = joblib.load('pd_model.pkl')
        pd_scaler = joblib.load('pd_scaler.pkl')
        pd_features = joblib.load('pd_features.pkl')
        print(" Loaded saved Parkinson's model.")
    except:
        print(" Training Parkinson's model on‑the‑fly...")
        url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/parkinsons/parkinsons.data'
        df_pd = pd.read_csv(url)
        df_pd.drop('name', axis=1, inplace=True)
        X = df_pd.drop('status', axis=1)
        y = df_pd['status']
        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X)
        model = RandomForestClassifier(n_estimators=100, random_state=42)
        model.fit(X_scaled, y)
        pd_model = model
        pd_scaler = scaler
        pd_features = list(X.columns)
        # joblib.dump(model, 'pd_model.pkl')
        # joblib.dump(scaler, 'pd_scaler.pkl')
        # joblib.dump(pd_features, 'pd_features.pkl')
        print(" Parkinson's model trained.")

    return (bc_model, bc_scaler, bc_features), (pd_model, pd_scaler, pd_features)

(bc_model, bc_scaler, bc_features), (pd_model, pd_scaler, pd_features) = get_models()

# -------------------------------
# 2. Prediction functions
# -------------------------------
def predict_breast_cancer():
    print("\n Breast Cancer Prediction (Malignant/Benign)")
    print("Enter the following 30 cell nuclei measurements (comma‑separated):")
    print("Feature order:", bc_features[:5], "... (total 30)")
    values = input("Enter all 30 values (comma‑separated): ")
    try:
        nums = [float(x.strip()) for x in values.split(',')]
        if len(nums) != len(bc_features):
            print(f" Expected {len(bc_features)} values, got {len(nums)}")
            return
        input_df = pd.DataFrame([nums], columns=bc_features)
        scaled = bc_scaler.transform(input_df)
        prob = bc_model.predict_proba(scaled)[0][1]  # probability of benign (1)
        pred = bc_model.predict(scaled)[0]
        result = "Benign (Non‑cancerous)" if pred == 1 else "Malignant (Cancerous)"
        confidence = prob if pred == 1 else 1 - prob
        print(f"\n Prediction: {result} (confidence {confidence:.2%})")
    except Exception as e:
        print(f"Error: {e}")

def predict_parkinson():
    print("\n Parkinson's Disease Prediction")
    print("Enter the following 22 voice measurements (comma‑separated):")
    print("Feature order:", pd_features[:5], "... (total 22)")
    values = input("Enter all 22 values (comma‑separated): ")
    try:
        nums = [float(x.strip()) for x in values.split(',')]
        if len(nums) != len(pd_features):
            print(f" Expected {len(pd_features)} values, got {len(nums)}")
            return
        input_df = pd.DataFrame([nums], columns=pd_features)
        scaled = pd_scaler.transform(input_df)
        prob = pd_model.predict_proba(scaled)[0][1]  # probability of PD (1)
        pred = pd_model.predict(scaled)[0]
        result = "Parkinson's Disease" if pred == 1 else "Healthy"
        confidence = prob if pred == 1 else 1 - prob
        print(f"\n Prediction: {result} (confidence {confidence:.2%})")
    except Exception as e:
        print(f"Error: {e}")

# -------------------------------
# 3. Interactive menu
# -------------------------------
print("\n" + "="*50)
print("DISEASE PREDICTION COMPARISON TOOL")
print("="*50)
print("Choose an option:")
print("1. Predict Breast Cancer (Malignant/Benign)")
print("2. Predict Parkinson's Disease (Healthy/PD)")
print("3. Exit")

while True:
    choice = input("\nEnter your choice (1/2/3): ").strip()
    if choice == '1':
        predict_breast_cancer()
    elif choice == '2':
        predict_parkinson()
    elif choice == '3':
        print("Exiting. Stay healthy!")
        break
    else:
        print("Invalid choice. Please enter 1, 2, or 3.")
```
